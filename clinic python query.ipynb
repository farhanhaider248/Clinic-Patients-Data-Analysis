```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("clinic unclean.csv")
```

```python
df.head()
```

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df['patientname'] = df['patientname'].str.strip().str.lower()
```

```python
df['gender'].value_counts()
```

```python
gen_map = {'male':'male','Female':'female','F':'female','Male':'male','M':'male','female':'female'}
df['gender'] = df['gender'].map(gen_map)
```

```python
df.head()
```

```python
df['doctor'] = df['doctor'].str.strip().str.lower()
df['department'] = df['department'].str.strip().str.lower()
df['insurancestatus'] = df['insurancestatus'].str.strip().str.lower()
df['status'] = df['status'].str.strip().str.lower()
```

```python
df.isnull().sum()
```

```python
df['age'] = df['age'].fillna(df['age'].median())
df['waittime_min'] = df['waittime_min'].fillna(df['waittime_min'].median())
```

```python
df.head(21)
```

```python
df.head(21)
```

```python
df['fee'] = df['fee'].fillna(
    df.groupby(['department','insurancestatus','status'])['fee'].transform('median'))
```

```python
df.isnull().sum()
```

```python
df['bloodpressure'] = df['bloodpressure'].fillna('unknown')
```
